Answer: **Indexes beating the US: 2 out of 10**

````python
import yfinance as yf
import pandas as pd
from datetime import date

end_date = date(2026, 8, 21)
periods = {'3y': 3, '5y': 5, '10y': 10}

# Descargamos suficiente histórico (10 años + margen) de una sola vez por índice
start_date = date(end_date.year - 11, end_date.month, end_date.day)

hist_data = {}
for country, info in common_world_index_tickers.items():
    hist = yf.Ticker(info['yahoo_ticker']).history(start=start_date.isoformat(), end=end_date.isoformat())
    hist.index = hist.index.tz_localize(None)   # el mismo ajuste que ya vimos
    hist_data[country] = hist['Close']

results = {}
for label, years in periods.items():
    period_start = pd.Timestamp(end_date.year - years, end_date.month, end_date.day)
    returns = {}
    for country, close in hist_data.items():
        start_close = close[close.index >= period_start].iloc[0]  # primer precio disponible desde esa fecha
        end_close = close.iloc[-1]                                 # último precio disponible
        returns[country] = end_close / start_close - 1
    results[label] = pd.Series(returns).sort_values(ascending=False)

for label, series in results.items():
    us = series['United States']
    beat = (series.drop('United States') > us).sum()
    print(f"{label}: US return = {us:.2%} -> {beat} out of 10 beat the US")
````

### Question 3. [Index] S&P 500 Market Corrections Analysis


**Calculate the median drawdown (in %) of significant market corrections in the S&P 500 index.**

For this task, define a correction as an event when a stock index goes down by **at least 5%** from the most recent all-time high.

Steps:
1. Download S&P 500 historical data (period from 1950 to present) using yfinance (daily stats)
2. Identify all-time high closing price points (where price exceeds all previous days' prices)
3. For each pair of consecutive all-time highs, find the minimum price in between
4. Calculate drawdown percentages: (high - low) / high × 100
5. Filter for corrections with at least 5% drawdown
6. Calculate the duration in days for each correction period
7. Determine the 25th, 50th (median), and 75th percentiles for correction durations and drawdowns

*Context:* 
> * Investors often wonder about the typical length of market corrections when deciding "when to buy the dip" ([Reddit discussion](https://www.reddit.com/r/investing/comments/1jrqnte/when_are_you_buying_the_dip/?rdt=64135)).

> * [A Wealth of Common Sense - How Often Should You Expect a Stock Market Correction?](https://awealthofcommonsense.com/2022/01/how-often-should-you-expect-a-stock-market-correction/)

*Hint (use this data to compare with your results)*: Here is the list of top 10 largest corrections by drawdown:
* 2007-10-09 to 2009-03-09: 56.8% drawdown over 517 days
* 2000-03-24 to 2002-10-09: 49.1% drawdown over 929 days
* 1973-01-11 to 1974-10-03: 48.2% drawdown over 630 days
* 1968-11-29 to 1970-05-26: 36.1% drawdown over 543 days
* 2020-02-19 to 2020-03-23: 33.9% drawdown over 33 days
* 1987-08-25 to 1987-12-04: 33.5% drawdown over 101 days
* 1961-12-12 to 1962-06-26: 28.0% drawdown over 196 days
* 1980-11-28 to 1982-08-12: 27.1% drawdown over 622 days
* 2022-01-03 to 2022-10-12: 25.4% drawdown over 282 days
* 1966-02-09 to 1966-10-07: 22.2% drawdown over 240 days

````python
import yfinance as yf
import pandas as pd
import numpy as np

# 1. Download S&P 500 historical data (period from 1950 to present) using yfinance (daily stats)

ticker_obj = yf.Ticker("^GSPC")
spx = ticker_obj.history(start="1950-01-01", interval="1d")
close = spx['Close']
close.index = close.index.tz_localize(None)

# 2. Identify all-time high closing price points (where price exceeds all previous days' prices)
cummax = close.cummax()

group = (cummax != cummax.shift()).cumsum()
df = pd.DataFrame({'Close': close, 'cummax': cummax, 'group': group})

# 3. For each pair of consecutive all-time highs, find the minimum price in between
# 4. Calculate drawdown percentages: (high - low) / high × 100
records = []
groups = list(df.groupby('group'))

for i, (gid, g) in enumerate(groups):
    peak_date, peak_value = g.index[0], g['cummax'].iloc[0]
    trough_date, trough_value = g['Close'].idxmin(), g['Close'].min()
    records.append({
        'peak_date': peak_date, 'peak_value': peak_value,
        'trough_date': trough_date, 'trough_value': trough_value,
        'drawdown_pct': (peak_value - trough_value) / peak_value * 100,
        'duration_days': (trough_date - peak_date).days,
        'is_last_group': i == len(groups) - 1
    })
drawdowns = pd.DataFrame(records)

resolved = drawdowns[~drawdowns['is_last_group']]

# 5. Filter for corrections with at least 5% drawdown
corrections = resolved[resolved['drawdown_pct'] >= 5].sort_values('drawdown_pct', ascending=False)

# 6. Calculate the duration in days for each correction period
# 7. Determine the 25th, 50th (median), and 75th percentiles for correction durations and drawdowns
for col in ['drawdown_pct', 'duration_days']:
    p25, p50, p75 = np.percentile(corrections[col], [25, 50, 75])
    print(f"{col}: p25={p25:.2f}  median={p50:.2f}  p75={p75:.2f}")
````

### Question 4.  [Stocks] Earnings Surprise Analysis for Amazon (AMZN)


**Calculate the median 2-day percentage change in stock prices following positive earnings surprise days.**

Steps:
1. Load earnings data for the last few years using the yfinance method get_earnings_dates():
```python
import yfinance as yf

ticker = 'AMZN'
ticker_obj = yf.Ticker(ticker)
result = ticker_obj.get_earnings_dates()
```
You should have 25 entries starting from 2020-10-29. Note: the 1 future earnings date entry will not have Reported EPS  
and Surprise % filled in.

2. Download complete historical price data using yfinance
3. Calculate 2-day percentage changes for all historical dates: for each sequence of 3 consecutive trading days (Day 1, Day 2, Day 3), compute the *return* as Close_Day3 / Close_Day1 - 1. (Assume Day 2 may correspond to the earnings announcement.)
4. Filter for positive earnings surprises and calculate the median 2-day return. Then calculate the correlation between
   the 2-day stock return and the earnings surprise magnitude.

▎ *Hint*: you can generate all correlations at once using pd.corr().

 > **Context:** Earnings announcements, especially when they exceed analyst expectations, can significantly impact stock prices in the short term.

**Reference:** [Yahoo Finance Earnings Calendar](https://finance.yahoo.com/calendar/earnings?symbol=AMZN)

*Additional*: Is there a correlation between the magnitude of the earnings surprise and the stock price reaction? Does the market react differently to earnings surprises during bull vs. bear markets?


````python
import yfinance as yf
import pandas as pd
import numpy as np

ticker_obj = yf.Ticker('AMZN')

# 1. Load earnings data for the last few years using the yfinance method get_earnings_dates()
earnings = ticker_obj.get_earnings_dates(limit=30)
earnings.index = earnings.index.tz_localize(None).normalize()
earnings = earnings[earnings.index >= '2020-10-29'].sort_index()

# 2. Download complete historical price data using yfinance
prices = ticker_obj.history(period="max", interval="1d")
prices.index = prices.index.tz_localize(None).normalize()
close = prices['Close']
trading_days = close.index

# 3. Calculate 2-day percentage changes for all historical dates: 
# for each sequence of 3 consecutive trading days (Day 1, Day 2, Day 3), 
# compute the *return* as Close_Day3 / Close_Day1 - 1. 
# (Assume Day 2 may correspond to the earnings announcement.)
records = []
for edate, row in earnings.iterrows():
    day2_pos = np.searchsorted(trading_days.values, np.datetime64(edate))
    day1_pos, day3_pos = day2_pos - 1, day2_pos + 1
    if day1_pos < 0 or day3_pos >= len(trading_days):
        continue
    records.append({
        'earnings_date': edate,
        'surprise_pct': row['Surprise(%)'],
        'ret_2day': close.iloc[day3_pos] / close.iloc[day1_pos] - 1
    })

df = pd.DataFrame(records).dropna(subset=['surprise_pct'])

# 4. Filter for positive earnings surprises and calculate the median 2-day return. 
# Then calculate the correlation between the 2-day stock return and the earnings 
# surprise magnitude.

positive_surprises = df[df['surprise_pct'] > 0]
median_ret = positive_surprises['ret_2day'].median()
corr = df[['ret_2day', 'surprise_pct']].corr()

print(f"Mediana retorno 2 días tras sorpresa positiva: {median_ret:.2%}")
print(corr)
````
